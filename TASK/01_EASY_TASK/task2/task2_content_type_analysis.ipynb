load the clean data set 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
os.makedirs("charts",exist_ok=(True))
```

```python
df=pd.read_csv("netflix_clean.csv", encoding="latin-1")
```

```python
df.head(2)
```

calculate the total number of movies and tv show

how much movie and the tv shows

```python
df = df.drop_duplicates()   # remove fully duplicate rows
df["type"].value_counts()
```

percentage provide ?

```python
type_count = df["type"].value_counts()
type_pct = (type_count / type_count.sum() * 100).round(2)
print(type_pct)
```

create visualization for content distribution 

```python
type_count = df["type"].value_counts()
type_pct = (type_count / type_count.sum() * 100).round(2)
fig, ax = plt.subplots(figsize=(5, 5))
ax.pie(type_count, labels=type_count.index, autopct="%1.1f%%",
       startangle=90, colors=["#E50914", "#EB834F"],
       wedgeprops={"width": 0.4, "edgecolor": "white"})
ax.set_title("Movies vs TV Shows")
plt.savefig("charts/task2_Movies_TV Shows.png", dpi=150)
plt.show()


```
