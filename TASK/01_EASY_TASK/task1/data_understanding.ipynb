```python
mask = df["rating"].astype(str).str.contains("min", na=False)
```

```python
# clean the text
for col in df.select_dtypes(include=["object", "str"]).columns:
    df[col] = df[col].str.strip()
```

```python
# handle missing values
for col in ["director", "cast", "country", "rating"]:
    if col in df.columns:
        df[col] = df[col].fillna("Unknown")
```

```python
print(df["date_added"].dtype)
```

```python
df["date_added"] = pd.to_datetime(df["date_added"], format="%m/%d/%Y")
```

```python
df["duration_minutes"] = df["duration"].str.extract(r"(\d+) min").astype(float)
```

```python
df["seasons"] = df["duration"].str.extract(r"(\d+) Season").astype(float)
```

```python
df["year_added"] = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month
df["years_to_add"] = df["year_added"] - df["release_year"]
df["main_country"] = df["country"].str.split(",").str[0].str.strip()
```
